```python
import requests
import random
import re
import math



def get_random_pfam(n_pages=25):

    url = "https://www.ebi.ac.uk/interpro/api/entry/pfam/?page_size=200"

    pfams = []

    for _ in range(n_pages):

        resp = requests.get(url)
        resp.raise_for_status()

        data = resp.json()

        for item in data["results"]:
            pfams.append(item["metadata"]["accession"])

        # next page URL
        url = data.get("next")

        if url is None:
            break

    return random.choice(pfams)



import requests
import time


def extract_pdb_ids(pfam_id):

    url = f"https://www.ebi.ac.uk/interpro/api/structure/pdb/entry/pfam/{pfam_id}"

    try:

        resp = requests.get(url, timeout=20)

        # HTTP error handling
        resp.raise_for_status()

        # Empty response
        if not resp.text.strip():
            print("EMPTY RESPONSE")
            return []

        # Non-json response protection
        try:
            data = resp.json()
        except Exception:
            print("INVALID JSON")
            print(resp.text[:200])
            return []

        pdb_ids = []

        for item in data.get("results", []):

            pdb = item["metadata"]["accession"]

            pdb_ids.append(pdb)

        return pdb_ids

    except requests.exceptions.RequestException as e:

        print("REQUEST FAILED:", e)

        return []
```

```python
i=0
# -----------------------------
while i < 3000:
    i+=1
    print(i)
    pfam_id = get_random_pfam()

    print("PFAM:", pfam_id)

    pdb_ids = extract_pdb_ids(pfam_id)

    print("PDB IDS:", pdb_ids)

    if pdb_ids != []:

        pdb_id = random.choice(pdb_ids)

        print("SELECTED:", pdb_id)

        resp = requests.get(
            url=f"https://files.rcsb.org/download/{pdb_id.upper()}.pdb"
        )

        out_pdb = f"PDB/{pfam_id}_extra.pdb"

        with open(out_pdb, "wb") as f:
            f.write(resp.content)

        print("Downloaded:", out_pdb)

    else:
        print("No PDB found for this PFAM")
    time.sleep(0.5)
```
